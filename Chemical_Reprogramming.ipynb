```python
chemicals = [
    [-8, 12, -6],
    [-5, -4, -9],
    [7, -3, -2]
]

changes = [10, -15, 10]

scores = []

for i in range(3):
     score = 0
     for j in range(3):
         if chemicals[i][j] * changes[j] < 0:
            score = score + 1
     scores.append(score)
print(scores)
```

```python
scores = [3, 2, 1]
chemical_names = ["X", "Y", "Z"]
highest_score = max(scores)
winner_index = scores.index(highest_score)

print(chemical_names[winner_index], highest_score)
```

```python
chemical_names = ["X", "Y", "Z"]
scores = [1, 3, 2]

chemical_pairs = list(zip(chemical_names, scores))
print(chemical_pairs)
```

```python
import pandas as pd

data = {
"Gene": ["GeneA", "GeneB", "GeneC"],
"Young": [10, 30, 5],
"Old": [20, 15, 12]
}

df = pd.DataFrame(data)

print(df)
```
