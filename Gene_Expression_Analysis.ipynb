```python
import pandas as pd
df = pd.read_csv("/content/GSE317925_raw_counts_all_samples.txt.gz", sep="\t")

print(df.head())
print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
controls = df[["ensembl_id", "Young_Mock_1", "Young_Mock_2", "Aged_Vector_1", "Aged_Vector_2"]]
print(controls.head())
```

```python
young_total_1 = controls["Young_Mock_1"].sum()
young_total_2 = controls["Young_Mock_2"].sum()
aged_total_1 = controls["Aged_Vector_1"].sum()
aged_total_2 = controls["Aged_Vector_2"].sum()

print(young_total_1)
print(young_total_2)
print(aged_total_1)
print(aged_total_2)
```

```python
controls["Young_Mock_1_CPM"] = controls["Young_Mock_1"] / young_total_1 * 1000000
print(controls.head())
```

```python
controls["Young_Mock_2_CPM"] = controls["Young_Mock_2"] / young_total_2 * 1000000
controls["Aged_Vector_1_CPM"] = controls["Aged_Vector_1"] / aged_total_1 * 1000000
controls["Aged_Vector_2_CPM"] = controls["Aged_Vector_2"] / aged_total_2 * 1000000
print(controls.head())
```

```python
controls["Young_Average_CPM"] = (controls["Young_Mock_1_CPM"] + controls["Young_Mock_2_CPM"]) / 2
print(controls.head())
```

```python
controls["Aged_Average_CPM"] = (controls["Aged_Vector_1_CPM"] + controls["Aged_Vector_2_CPM"]) / 2
print(controls.head())
```

```python
controls["Fold_Change"] = controls["Aged_Average_CPM"] / controls["Young_Average_CPM"]

print(controls[["ensembl_id", "Young_Average_CPM", "Aged_Average_CPM", "Fold_Change"]].head())
```

```python
controls["Adjusted_Fold_Change"] = (controls["Aged_Average_CPM"] + 1) / (controls["Young_Average_CPM"] + 1)
print(controls[["ensembl_id", "Adjusted_Fold_Change"]].head())
```

```python
import numpy as np

controls["Log2_Fold_Change"] = np.log2(controls["Adjusted_Fold_Change"])

print(controls[["ensembl_id", "Log2_Fold_Change"]].head())
```
